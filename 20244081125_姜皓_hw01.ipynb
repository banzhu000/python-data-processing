```python
def gcd(m, n):
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m

print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
print(gcd(-12, -8))
```

```python
import math

def euler_limit(n):
    return (1 + 1/n)**n

def euler_infinite_sum(n):
    # 前n项级数和 k=0 ~ n-1
    if n == 0:
        return 0.0
    total = 1.0
    fact = 1
    for k in range(1, n):
        fact *= k
        total += 1 / fact
    return total

def euler_approx(epsilon):
    total = 1.0
    fact = 1
    k = 1
    while True:
        term = 1 / fact
        if term < epsilon:
            break
        total += term
        k +=1
        fact *=k
    return total

def print_euler_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")

#测试
print(euler_limit(10000))
print("真实e：", math.e)
print(euler_infinite_sum(1))
print(euler_infinite_sum(4))
print(euler_approx(1e-5))
print_euler_table(6)

```

```python
def is_power_of_two(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

def is_power(b, n):
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

#测试
print(is_power_of_two(1))
print(is_power_of_two(8))
print(is_power_of_two(10))
print(is_power(3, 9))
print(is_power(2, 16))
print(is_power(5,12))
```

```python
import os
import pickle

#1 创建目录
cwd = os.getcwd()
print("当前工作目录",cwd)
data_dir = os.path.join(cwd,"exp01_data")
os.makedirs(data_dir,exist_ok=True)

#准备students.txt文本内容
txt_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

stu_file = os.path.join(data_dir,"students.txt")
#2写入再读取
with open(stu_file,"w",encoding="utf-8") as f:
    f.write(txt_content)

#导入外部模块
from student_utils import parse_student

valid_records = []
error_info = []

with open(stu_file,"r",encoding="utf-8") as f:
    lines = f.readlines()
for lineno,line in enumerate(lines,start=1):
    if lineno ==1:
        continue #跳过表头
    try:
        res = parse_student(line)
        if res is not None:
            valid_records.append(res)
    except ValueError as e:
        error_info.append((lineno,str(e)))

print("有效记录：",valid_records)
print("错误行：",error_info)

#统计
valid_num = len(valid_records)
invalid_num = len(error_info)
if valid_num>0:
    avg = sum(r["score"] for r in valid_records)/valid_num
    pass_count = sum(1 for r in valid_records if r["score"]>=60)
    pass_rate = pass_count / valid_num
else:
    avg=0
    pass_rate=0

report_path = os.path.join(data_dir,"report.txt")
# w模式覆盖写
with open(report_path,"w",encoding="utf-8") as f:
    f.write(f"有效记录数:{valid_num}\n")
    f.write(f"无效记录数:{invalid_num}\n")
    f.write(f"平均分:{avg:.2f}\n")
    f.write(f"及格率:{pass_rate:.1%}\n")

# a追加模式演示
with open(report_path,"a",encoding="utf-8") as f:
    f.write("【追加测试行】\n")

#pickle保存读取
pkl_path = os.path.join(data_dir,"students.pkl")
with open(pkl_path,"wb") as f:
    pickle.dump(valid_records,f)

with open(pkl_path,"rb") as f:
    loaded = pickle.load(f)
print("pickle读取是否相等", loaded == valid_records)

#测试parse_student函数
print("有效测试",parse_student("Tom,77"))
try:
    parse_student("Tom,200")
except ValueError as e:
    print("无效测试报错",e)

#捕获FileNotFoundError
try:
    with open("not_exist_1234.txt","r",encoding="utf-8") as f:
        pass
except FileNotFoundError as e:
    print(f"捕获文件不存在异常：{e}")
```
