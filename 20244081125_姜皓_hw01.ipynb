```python
def gcd(m, n):
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m

print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
print(gcd(-12, -8))
```

```python
def euler_limit(n):
    return (1 + 1/n)**n

def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1
    for k in range(1, n):
        fact *= k
        total += 1/fact
    return total

def euler_approx(epsilon):
    total = 0.0
    fact = 1
    k = 1
    while True:
        term = 1 / fact
        if term < epsilon:
            break
        total += term
        k +=1
        fact *=k
    return total

def print_euler_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")

#测试
print(euler_limit(10000))
print("真实e：", math.e)
print(euler_infinite_sum(1))
print(euler_infinite_sum(4))
print(euler_approx(1e-5))
print_euler_table(6)
```

```python
def is_power_of_two(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

def is_power(b, n):
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

#测试
print(is_power_of_two(1))
print(is_power_of_two(8))
print(is_power_of_two(10))
print(is_power(3, 9))
print(is_power(2, 16))
print(is_power(5,12))
```
