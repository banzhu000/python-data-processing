```python
# 1. 回文判断
def is_palindrome(s):
    s = ''.join(filter(str.isalnum, s.lower()))
    return s == s[::-1]

# 测试用例
print(is_palindrome("A man, a plan, a canal, Panama"))  # True
print(is_palindrome("race a car"))  # False
print(is_palindrome(" "))
```

```python
# 2. 字母顺序判断
def is_abecedarian(s):
    s = ''.join(filter(str.isalpha, s.lower()))
    return s == ''.join(sorted(s))

# 测试用例
print(is_abecedarian("abc"))  # True
print(is_abecedarian("aBc"))  # True
print(is_abecedarian("abc def"))  # True
print(is_abecedarian("abc def ghi"))  # True
print(is_abecedarian("xyz"))  # True
print(is_abecedarian("xyz abc"))  # False
print(is_abecedarian("xyz abc def"))  # False
```

```python
# 3. 去除元音
def remove_vowels(s):
    vowels = 'aeiou'
    return ''.join(ch for ch in s if ch not in vowels)

# 测试用例
print(remove_vowels("abcdef"))  # "bcdf"
print(remove_vowels("abcdefg"))  # "bcdfg"
print(remove_vowels("ACED"))  # "CD"
print(remove_vowels("AEIOU"))  # ""
print(remove_vowels("aAeEiIoOuU"))  # ""
```

```python
# 4. 列表反转
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]

# 测试用例
print(list_reverse([1, 2, 3]))  # [3, 2, 1]
print(list_reverse([1, 2]))  # [2, 1]
print(list_reverse([]))  # []
try:
    print(list_reverse("abc"))  # 应该引发 TypeError
except TypeError as e:
    print(e)
```

```python
# 5. 列表排序判断
def is_sorted(seq):
    return all(seq[i] <= seq[i + 1] for i in range(len(seq) - 1))

# 测试用例
print(is_sorted([1, 2, 3]))  # True
print(is_sorted([1, 3, 2]))  # False
print(is_sorted([]))  # True
print(is_sorted([5, 5, 5]))  # True
```

```python
# 6. 向量标量乘法
def vec_scalar_mult(t, s):
    if not (isinstance(t, tuple) and all(isinstance(x, (int, float)) for x in t)):
        raise TypeError("输入必须是整数或浮点数组成的元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或浮点数")
    return tuple(x * s for x in t)

# 测试用例
print(vec_scalar_mult((1, 2, 3), 2))  # (2, 4, 6)
print(vec_scalar_mult((1, -2, 3), -1))  # (-1, 2, -3)
try:
    print(vec_scalar_mult((1, 2, 3), "a"))  # 应该引发 TypeError
except TypeError as e:
    print(e)
```

```python
# 7. 向量内积
def vec_inner_product(x, y):
    if not (isinstance(x, tuple) and isinstance(y, tuple)):
        raise TypeError("输入必须是元组")
    if len(x) != len(y):
        raise ValueError("两个元组的长度必须相同")
    return sum(xi * yi for xi, yi in zip(x, y))

# 测试用例
print(vec_inner_product((1, 2, 3), (4, 5, 6)))  # 32
print(vec_inner_product((-1, -2, -3), (-4, -5, -6)))  # -32
try:
    print(vec_inner_product((1, 2, 3), (4, 5)))  # 应该引发 ValueError
except ValueError as e:
    print(e)
```

```python
# 8. 矩阵有效性检查
def check_valid_mx(my_mx):
    if not (isinstance(my_mx, tuple) and all(isinstance(row, tuple) for row in my_mx)):
        return False
    if not all(len(row) == len(my_mx[0]) for row in my_mx):
        return False
    if not all(isinstance(x, (int, float)) for row in my_mx for x in row):
        return False
    return True

# 测试用例
print(check_valid_mx(((1, 2), (3, 4))))  # True
print(check_valid_mx(((1, 2), (3, 4, 5))))  # False
print(check_valid_mx(((1, 2), (3, "4"))))  # False
print(check_valid_mx(((1, 2), (3, 4.0))))  # True
```

```python
# 9. 矩阵向量乘法
def mx_vec_mult(my_mx, v):
    if not check_valid_mx(my_mx):
        raise ValueError("无效的矩阵")
    if not isinstance(v, tuple) or len(v) != len(my_mx[0]):
        raise ValueError("无效的向量")
    return tuple(sum(my_mx[i][j] * v[j] for j in range(len(v))) for i in range(len(my_mx)))

# 测试用例
print(mx_vec_mult(((1, 2), (3, 4)), (1, 2)))  # (5, 11)
print(mx_vec_mult(((1, 2), (3, 4)), (0, 1)))  # (2, 4)
try:
    print(mx_vec_mult(((1, 2), (3, 4)), (1,)))  # 应该引发 ValueError
except ValueError as e:
    print(e)
```

```python
# 10. 向量类
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise ValueError("维度必须是非负整数")
        if elements is None:
            self.elements = (0,) * dim
        elif isinstance(elements, (list, tuple)) and len(elements) == dim:
            self.elements = tuple(elements)
        else:
            raise ValueError("元素必须与维度匹配")

    @property
    def dim(self):
        return len(self.elements)

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements >= other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是Vector对象")
        if self.dim != other.dim:
            raise ValueError("维度不同，无法计算内积")
        return sum(x * y for x, y in zip(self.elements, other.elements))

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return Vector(self.dim, tuple(x * other for x in self.elements))
        elif isinstance(other, Vector):
            return self.dot(other)
        else:
            raise TypeError("不支持的运算")

    __rmul__ = __mul__

# 测试用例
v1 = Vector(3, (1, 2, 3))
v2 = Vector(3, (4, 5, 6))
v3 = Vector(2, (1, 2))

print(v1 == v1)  # True
print(v1 < v2)  # True
print(v1 <= v2)  # True
print(v1 > v2)  # False
print(v1 >= v2)  # False
print(v1.dot(v2))  # 32
print(v1 * 2)  # Vector(3, (2, 4, 6))
print(2 * v1)  # Vector(3, (2, 4, 6))
try:
    v1 + v2  # 应该引发 TypeError
except TypeError as e:
    print(e)
```
